```python
# Load the three phishing datasets into pandas
# UCI is an .arff file, the other two are CSV
import pandas as pd
import numpy as np
from scipy.io import arff

# UCI (.arff)
data, meta = arff.loadarff('Training Dataset.arff')
uci = pd.DataFrame(data)

# Hannousse (.csv)
hannousse = pd.read_csv('dataset_B_05_2020.csv')

# Vrbancic (.csv)
vrbancic = pd.read_csv('dataset_full.csv')

print("UCI:", uci.shape)
print("Hannousse:", hannousse.shape)
print("Vrbancic:", vrbancic.shape)

# --- added for screenshot 1: preview each dataset ---
display(uci.head())
display(hannousse.head())
display(vrbancic.head())
```

```python
# Remove duplicate rows from Hannousse and Vrbancic
data, meta = arff.loadarff('Training Dataset.arff')
uci = pd.DataFrame(data)

# UCI: keep all rows. Its 30 discretised features mean identical rows are
# coincidental feature patterns, not true duplicate websites.
# Only remove duplicates from the richer datasets:
hannousse = hannousse.drop_duplicates().reset_index(drop=True)
vrbancic = vrbancic.drop_duplicates().reset_index(drop=True)

print("UCI (kept full):", uci.shape)
print("Hannousse:", hannousse.shape)
print("Vrbancic:", vrbancic.shape)
```

```python
# UCI has only 30 features with values -1, 0, 1, so many rows look identical
# but are actually different websites. I keep all UCI rows and only remove
# real duplicates from the other two datasets.

print("UCI columns (last 3):", list(uci.columns[-3:]))
print("Hannousse columns (first 2 + last 2):", list(hannousse.columns[:2]) + list(hannousse.columns[-2:]))
print("Vrbancic columns (first 2 + last 2):", list(vrbancic.columns[:2]) + list(vrbancic.columns[-2:]))

print("\n--- Class balance ---")
print("\nUCI (last column):")
print(uci.iloc[:, -1].value_counts())

print("\nHannousse (status column):")
print(hannousse['status'].value_counts() if 'status' in hannousse.columns else "check column name above")

print("\nVrbancic (phishing column):")
print(vrbancic['phishing'].value_counts() if 'phishing' in vrbancic.columns else "check column name above")
```

```python
# The three datasets label phishing differently (1/-1, text, 0/1).
# I convert them all to the same format: 0 = legitimate, 1 = phishing

# UCI: Result is b'1' (phishing) / b'-1' (legitimate), stored as bytes
uci['label'] = uci['Result'].apply(lambda x: 1 if x == b'1' else 0)

# Hannousse: status is 'phishing' / 'legitimate'
hannousse['label'] = hannousse['status'].apply(lambda x: 1 if x == 'phishing' else 0)

# Vrbancic: phishing is already 1 (phishing) / 0 (legitimate)
vrbancic['label'] = vrbancic['phishing']

# Confirm all three now use the same 0/1 format
print("UCI label:\n", uci['label'].value_counts())
print("\nHannousse label:\n", hannousse['label'].value_counts())
print("\nVrbancic label:\n", vrbancic['label'].value_counts())
```

```python
# Step 5b: Group features into URL / CONTENT / DOMAIN categories

uci_url = ['having_IP_Address','URL_Length','Shortining_Service','having_At_Symbol',
           'double_slash_redirecting','Prefix_Suffix','having_Sub_Domain','port','HTTPS_token']
uci_content = ['Favicon','Request_URL','URL_of_Anchor','Links_in_tags','SFH',
               'Submitting_to_email','Redirect','on_mouseover','RightClick','popUpWidnow','Iframe']
uci_domain = ['SSLfinal_State','Domain_registeration_length','Abnormal_URL','age_of_domain',
              'DNSRecord','web_traffic','Page_Rank','Google_Index','Links_pointing_to_page','Statistical_report']
uci_drop = ['Result']

hann_content = ['nb_hyperlinks','ratio_intHyperlinks','ratio_extHyperlinks','ratio_nullHyperlinks',
                'nb_extCSS','ratio_intRedirection','ratio_extRedirection','ratio_intErrors','ratio_extErrors',
                'login_form','external_favicon','links_in_tags','submit_email','ratio_intMedia','ratio_extMedia',
                'sfh','iframe','popup_window','safe_anchor','onmouseover','right_clic','empty_title',
                'domain_in_title','domain_with_copyright']
hann_domain = ['whois_registered_domain','domain_registration_length','domain_age','web_traffic',
               'dns_record','google_index','page_rank']
hann_drop = ['url','status']
hann_url = [c for c in hannousse.columns if c not in hann_content + hann_domain + hann_drop + ['label']]

vrb_domain = ['time_response','domain_spf','asn_ip','time_domain_activation','time_domain_expiration',
              'qty_ip_resolved','qty_nameservers','qty_mx_servers','ttl_hostname','tls_ssl_certificate',
              'qty_redirects','url_google_index','domain_google_index']
vrb_content = []
vrb_drop = ['phishing']
vrb_url = [c for c in vrbancic.columns if c not in vrb_domain + vrb_drop + ['label']]

categories = {
    'UCI':       {'url': uci_url,  'content': uci_content,  'domain': uci_domain,  'df': uci,       'drop': uci_drop},
    'Hannousse': {'url': hann_url, 'content': hann_content, 'domain': hann_domain, 'df': hannousse, 'drop': hann_drop},
    'Vrbancic':  {'url': vrb_url,  'content': vrb_content,  'domain': vrb_domain,  'df': vrbancic,  'drop': vrb_drop},
}

print('SAFETY CHECK')
for name, c in categories.items():
    assigned = set(c['url']) | set(c['content']) | set(c['domain']) | set(c['drop']) | {'label'}
    allcols = set(c['df'].columns)
    missing = sorted(allcols - assigned)
    unknown = sorted(assigned - allcols)
    print('  ', name, 'missing=', missing, 'unknown=', unknown)

print()
print('TABLE 1 - Feature-category mapping')
print('Dataset       URL   Content   Domain   Total')
for name, c in categories.items():
    u, ct, d = len(c['url']), len(c['content']), len(c['domain'])
    print('%-12s %4d %8d %8d %7d' % (name, u, ct, d, u + ct + d))
```

```python
# Step 6 (scalable): within-dataset baseline, 5 models, stratified 5-fold CV
import pandas as pd, numpy as np
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import StratifiedKFold, cross_validate
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, HistGradientBoostingClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import LinearSVC

def clean_X(X):
    return X.apply(lambda s: pd.to_numeric(
        s.map(lambda v: v.decode() if isinstance(v, bytes) else v), errors='coerce'))

models = {
    'Decision Tree':     DecisionTreeClassifier(random_state=42),
    'Random Forest':     RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1),
    'KNN':               KNeighborsClassifier(n_neighbors=5, n_jobs=-1),
    'SVM':               LinearSVC(dual='auto', max_iter=5000, random_state=42),
    'Gradient Boosting': HistGradientBoostingClassifier(random_state=42),
}
scoring = ['accuracy','precision','recall','f1','roc_auc']
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

baseline_rows = []
for name, c in categories.items():
    feats = c['url'] + c['content'] + c['domain']
    X = clean_X(c['df'][feats]).fillna(0)
    y = c['df']['label'].astype(int)
    print('Running', name, '...', X.shape)
    for mname, model in models.items():
        pipe = make_pipeline(StandardScaler(), model)
        res = cross_validate(pipe, X, y, cv=cv, scoring=scoring, n_jobs=-1)
        baseline_rows.append({
            'Dataset': name, 'Model': mname,
            'Accuracy':  res['test_accuracy'].mean(),
            'Precision': res['test_precision'].mean(),
            'Recall':    res['test_recall'].mean(),
            'F1':        res['test_f1'].mean(),
            'ROC_AUC':   res['test_roc_auc'].mean(),
        })
        print('   done:', mname)

baseline = pd.DataFrame(baseline_rows)
pd.set_option('display.float_format', lambda v: '%.4f' % v)
print('\nTABLE 2 - Within-dataset baseline (mean over 5 folds)')
print(baseline.to_string(index=False))
baseline.to_csv('baseline_results.csv', index=False)
```

```python
# Step 7a: verify label direction is consistent (1 should = phishing everywhere)
checks = {
    'UCI':       ['having_IP_Address'],      # having an IP in the URL = phishing signal
    'Hannousse': ['ip', 'google_index'],     # ip=1 phishing-ish; google_index=1 legit-ish
    'Vrbancic':  ['domain_in_ip', 'qty_at_url'],  # ip in domain / '@' in url = phishing signals
}
for name, feats in checks.items():
    df = categories[name]['df']
    print('='*55)
    print(name, '| phishing rate (mean label) =', round(df['label'].mean(), 3))
    for f in feats:
        s = df[f].map(lambda v: v.decode() if isinstance(v, bytes) else v)
        s = pd.to_numeric(s, errors='coerce')
        g = s.groupby(df['label']).mean()
        print(f'   {f:20} mean when label=0: {g.get(0, float("nan")):.3f} | label=1: {g.get(1, float("nan")):.3f}')
```

```python
# Step 8: harmonised cross-dataset test (Random Forest, full common feature set)
import numpy as np, pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, recall_score, f1_score

# canonical features: name, category, {dataset: column or None}
CANON = [
 ('f_ip','url',       {'UCI':'having_IP_Address','Hannousse':'ip','Vrbancic':'domain_in_ip'}),
 ('f_at','url',       {'UCI':'having_At_Symbol','Hannousse':'nb_at','Vrbancic':'qty_at_url'}),
 ('f_hyphen','url',   {'UCI':'Prefix_Suffix','Hannousse':'nb_hyphens','Vrbancic':'qty_hyphen_url'}),
 ('f_shorten','url',  {'UCI':'Shortining_Service','Hannousse':'shortening_service','Vrbancic':'url_shortened'}),
 ('f_urllen','url',   {'UCI':'URL_Length','Hannousse':'length_url','Vrbancic':'length_url'}),
 ('f_subdom','url',   {'UCI':'having_Sub_Domain','Hannousse':'nb_subdomains','Vrbancic':'qty_dot_domain'}),
 ('f_dns','domain',   {'UCI':'DNSRecord','Hannousse':'dns_record','Vrbancic':'qty_nameservers'}),
 ('f_gindex','domain',{'UCI':'Google_Index','Hannousse':'google_index','Vrbancic':'domain_google_index'}),
 ('f_domage','domain',{'UCI':'age_of_domain','Hannousse':'domain_age','Vrbancic':'time_domain_activation'}),
 ('f_domreg','domain',{'UCI':'Domain_registeration_length','Hannousse':'domain_registration_length','Vrbancic':'time_domain_expiration'}),
 ('f_iframe','content',{'UCI':'Iframe','Hannousse':'iframe','Vrbancic':None}),
 ('f_sfh','content',  {'UCI':'SFH','Hannousse':'sfh','Vrbancic':None}),
 ('f_anchor','content',{'UCI':'URL_of_Anchor','Hannousse':'safe_anchor','Vrbancic':None}),
 ('f_linktags','content',{'UCI':'Links_in_tags','Hannousse':'links_in_tags','Vrbancic':None}),
 ('f_submail','content',{'UCI':'Submitting_to_email','Hannousse':'submit_email','Vrbancic':None}),
]
canon_cat = {n: cat for n, cat, _ in CANON}

def to_num(s):
    return pd.to_numeric(s.map(lambda v: v.decode() if isinstance(v, bytes) else v), errors='coerce')

def build_bin(name):
    df = categories[name]['df']
    out = pd.DataFrame(index=df.index)
    for n, cat, mp in CANON:
        col = mp[name]
        if col is None:
            continue
        s = to_num(df[col]).fillna(0)
        out[n] = (s == s.max()).astype(int) if s.nunique() <= 3 else (s > s.median()).astype(int)
    out['label'] = df['label'].astype(int).values
    return out

BIN = {n: build_bin(n) for n in categories}

def orient(Xtr, ytr):
    signs = {}
    yv = ytr.values
    for col in Xtr.columns:
        xv = Xtr[col].values
        c = np.corrcoef(xv, yv)[0, 1] if xv.std() > 0 else 0
        signs[col] = -1 if (c == c and c < 0) else 1
    return signs

def apply_signs(X, signs):
    X = X.copy()
    for col, sg in signs.items():
        if sg < 0:
            X[col] = 1 - X[col]
    return X

names = ['UCI', 'Hannousse', 'Vrbancic']
rf = lambda: RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1)

print('CROSS-DATASET ACCURACY (Random Forest, full common feature set)')
print('rows = TRAIN on, cols = TEST on\n')
header = 'train\\test  ' + ''.join('%-12s' % n for n in names)
print(header)
acc_matrix = {}
for A in names:
    rowvals = []
    for B in names:
        feats = [c for c in BIN[A].columns if c != 'label' and c in BIN[B].columns]
        if A == B:
            Xtr, Xte, ytr, yte = train_test_split(
                BIN[A][feats], BIN[A]['label'], test_size=0.2,
                stratify=BIN[A]['label'], random_state=42)
        else:
            Xtr, ytr = BIN[A][feats], BIN[A]['label']
            Xte, yte = BIN[B][feats], BIN[B]['label']
        signs = orient(Xtr, ytr)
        m = rf().fit(apply_signs(Xtr, signs), ytr)
        pred = m.predict(apply_signs(Xte, signs))
        a = accuracy_score(yte, pred)
        acc_matrix[(A, B)] = (a, recall_score(yte, pred), f1_score(yte, pred), len(feats))
        rowvals.append(a)
    print('%-11s ' % A + ''.join('%-12.3f' % v for v in rowvals))

print('\nDetail (accuracy / recall / F1 / #features):')
for A in names:
    for B in names:
        a, r, f, nf = acc_matrix[(A, B)]
        tag = 'within' if A == B else 'CROSS'
        print('  %-9s -> %-9s [%s]  acc=%.3f  rec=%.3f  f1=%.3f  (%d feats)' % (A, B, tag, a, r, f, nf))
```

```python
# Step 9: per-category cross-dataset transfer, all 5 models
import pandas as pd, numpy as np
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, recall_score, f1_score
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, HistGradientBoostingClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import LinearSVC

model_factories = {
    'Decision Tree':     lambda: DecisionTreeClassifier(random_state=42),
    'Random Forest':     lambda: RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1),
    'KNN':               lambda: KNeighborsClassifier(n_neighbors=5, n_jobs=-1),
    'SVM':               lambda: LinearSVC(dual='auto', max_iter=5000, random_state=42),
    'Gradient Boosting': lambda: HistGradientBoostingClassifier(random_state=42),
}
cat_feats = {
    'all':     [n for n in canon_cat],
    'url':     [n for n, c in canon_cat.items() if c == 'url'],
    'content': [n for n, c in canon_cat.items() if c == 'content'],
    'domain':  [n for n, c in canon_cat.items() if c == 'domain'],
}
names = ['UCI', 'Hannousse', 'Vrbancic']

rows = []
for cat, feat_pool in cat_feats.items():
    for A in names:
        for B in names:
            feats = [f for f in feat_pool if f in BIN[A].columns and f in BIN[B].columns]
            if len(feats) == 0:
                continue   # e.g. content for Vrbancic
            if A == B:
                Xtr, Xte, ytr, yte = train_test_split(
                    BIN[A][feats], BIN[A]['label'], test_size=0.2,
                    stratify=BIN[A]['label'], random_state=42)
            else:
                Xtr, ytr = BIN[A][feats], BIN[A]['label']
                Xte, yte = BIN[B][feats], BIN[B]['label']
            signs = orient(Xtr, ytr)
            Xtr2, Xte2 = apply_signs(Xtr, signs), apply_signs(Xte, signs)
            for mname, factory in model_factories.items():
                m = factory().fit(Xtr2, ytr)
                pred = m.predict(Xte2)
                rows.append({
                    'category': cat, 'train': A, 'test': B, 'model': mname,
                    'type': 'within' if A == B else 'cross',
                    'n_feats': len(feats),
                    'accuracy': accuracy_score(yte, pred),
                    'recall':   recall_score(yte, pred, zero_division=0),
                    'f1':       f1_score(yte, pred, zero_division=0),
                })

cross = pd.DataFrame(rows)
cross.to_csv('crossdataset_by_category.csv', index=False)

# Summary: per category, mean within vs mean cross accuracy, and the drop
print('TABLE 3 - Cross-dataset transfer by feature category (mean over models & dataset pairs)')
print('%-9s %8s %8s %8s' % ('Category', 'Within', 'Cross', 'Drop'))
for cat in ['all', 'url', 'content', 'domain']:
    sub = cross[cross.category == cat]
    w = sub[sub.type == 'within']['accuracy'].mean()
    c = sub[sub.type == 'cross']['accuracy'].mean()
    print('%-9s %8.3f %8.3f %8.3f' % (cat, w, c, w - c))
```

```python
# Step 11: SHAP + permutation importance, aggregated by category
!pip install shap -q
import shap, numpy as np, pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.inspection import permutation_importance
from sklearn.model_selection import train_test_split

names = ['UCI', 'Hannousse', 'Vrbancic']

def category_importance(name):
    feats = [c for c in BIN[name].columns if c != 'label']
    X = apply_signs(BIN[name][feats], orient(BIN[name][feats], BIN[name]['label']))
    y = BIN[name]['label']
    Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)
    rf = RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1).fit(Xtr, ytr)

    # --- SHAP (sampled for speed) ---
    Xs = Xte.sample(min(1500, len(Xte)), random_state=42)
    sv = shap.TreeExplainer(rf).shap_values(Xs)
    if isinstance(sv, list):        sv = sv[1]
    elif getattr(sv, 'ndim', 2) == 3: sv = sv[:, :, 1]
    shap_imp = np.abs(sv).mean(axis=0)

    # --- permutation importance ---
    perm = permutation_importance(rf, Xte, yte, n_repeats=5, random_state=42, n_jobs=-1).importances_mean

    df = pd.DataFrame({'feature': feats, 'category': [canon_cat[f] for f in feats],
                       'shap': shap_imp, 'perm': perm})
    # aggregate to category, normalise to proportion of total importance
    agg = df.groupby('category')[['shap', 'perm']].sum()
    agg = agg.clip(lower=0)
    agg = agg / agg.sum()
    return agg

print('CATEGORY IMPORTANCE  (share of total importance, by method)')
for name in names:
    agg = category_importance(name)
    print('\n' + '='*45 + '\n' + name)
    print('%-9s %8s %8s' % ('Category', 'SHAP', 'Perm'))
    for cat in ['url', 'content', 'domain']:
        if cat in agg.index:
            print('%-9s %8.3f %8.3f' % (cat, agg.loc[cat, 'shap'], agg.loc[cat, 'perm']))
```

```python
# Step 12: statistical significance tests
from scipy.stats import wilcoxon, friedmanchisquare
from statsmodels.stats.contingency_tables import mcnemar
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
import numpy as np, pandas as pd

model_cols = ['Decision Tree','Random Forest','KNN','SVM','Gradient Boosting']

# --- Test A: is cross-dataset significantly lower than within? (Wilcoxon, paired by category+model) ---
w_vals, c_vals = [], []
for cat in ['all','url','content','domain']:
    for mdl in model_cols:
        w = cross[(cross.category==cat)&(cross.type=='within')&(cross.model==mdl)]['accuracy'].mean()
        c = cross[(cross.category==cat)&(cross.type=='cross')&(cross.model==mdl)]['accuracy'].mean()
        if pd.notna(w) and pd.notna(c):
            w_vals.append(w); c_vals.append(c)
stat, p = wilcoxon(w_vals, c_vals)
print('TEST A - Within vs Cross (Wilcoxon signed-rank)')
print('   mean within=%.3f  mean cross=%.3f  p=%.5f  -> %s' %
      (np.mean(w_vals), np.mean(c_vals), p, 'SIGNIFICANT' if p<0.05 else 'not significant'))

# --- Test B: do the 5 models differ on cross-dataset accuracy? (Friedman) ---
piv = (cross[cross.type=='cross']
       .pivot_table(index=['category','train','test'], columns='model', values='accuracy')
       .dropna())
stat, p = friedmanchisquare(*[piv[m] for m in model_cols])
print('\nTEST B - Model differences, cross-dataset (Friedman)')
print('   p=%.5f  -> %s' % (p, 'models differ significantly' if p<0.05 else 'no significant difference'))

# --- Test C: within vs cross on the SAME test samples (McNemar), UCI->Hannousse case ---
feats = [c for c in BIN['Hannousse'].columns if c!='label' and c in BIN['UCI'].columns]
Xtr_h, Xte_h, ytr_h, yte_h = train_test_split(
    BIN['Hannousse'][feats], BIN['Hannousse']['label'], test_size=0.2,
    stratify=BIN['Hannousse']['label'], random_state=42)
s_in = orient(Xtr_h, ytr_h)
m_in = RandomForestClassifier(100, random_state=42, n_jobs=-1).fit(apply_signs(Xtr_h, s_in), ytr_h)
pred_in = m_in.predict(apply_signs(Xte_h, s_in))
s_cr = orient(BIN['UCI'][feats], BIN['UCI']['label'])
m_cr = RandomForestClassifier(100, random_state=42, n_jobs=-1).fit(apply_signs(BIN['UCI'][feats], s_cr), BIN['UCI']['label'])
pred_cr = m_cr.predict(apply_signs(Xte_h, s_cr))
ok_in = (pred_in == yte_h.values); ok_cr = (pred_cr == yte_h.values)
table = [[np.sum(ok_in & ok_cr),  np.sum(ok_in & ~ok_cr)],
         [np.sum(~ok_in & ok_cr), np.sum(~ok_in & ~ok_cr)]]
res = mcnemar(table, exact=False, correction=True)
print('\nTEST C - Within(Hannousse) vs Cross(UC->Hann) on same samples (McNemar)')
print('   within acc=%.3f  cross acc=%.3f  p=%.5f  -> %s' %
      (ok_in.mean(), ok_cr.mean(), res.pvalue, 'SIGNIFICANT' if res.pvalue<0.05 else 'not significant'))
```

```python
# Verify McNemar: print the discordant counts and recompute
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from statsmodels.stats.contingency_tables import mcnemar
import numpy as np

feats = [c for c in BIN['Hannousse'].columns if c!='label' and c in BIN['UCI'].columns]
Xtr_h, Xte_h, ytr_h, yte_h = train_test_split(
    BIN['Hannousse'][feats], BIN['Hannousse']['label'], test_size=0.2,
    stratify=BIN['Hannousse']['label'], random_state=42)
s_in = orient(Xtr_h, ytr_h)
m_in = RandomForestClassifier(100, random_state=42, n_jobs=-1).fit(apply_signs(Xtr_h, s_in), ytr_h)
pred_in = m_in.predict(apply_signs(Xte_h, s_in))
s_cr = orient(BIN['UCI'][feats], BIN['UCI']['label'])
m_cr = RandomForestClassifier(100, random_state=42, n_jobs=-1).fit(apply_signs(BIN['UCI'][feats], s_cr), BIN['UCI']['label'])
pred_cr = m_cr.predict(apply_signs(Xte_h, s_cr))

ok_in = (pred_in == yte_h.values); ok_cr = (pred_cr == yte_h.values)
both_ok   = int(np.sum(ok_in & ok_cr))
in_only   = int(np.sum(ok_in & ~ok_cr))   # b: within correct, cross wrong
cr_only   = int(np.sum(~ok_in & ok_cr))   # c: cross correct, within wrong
both_wrong= int(np.sum(~ok_in & ~ok_cr))
print('Test samples:', len(yte_h))
print('Both correct      :', both_ok)
print('Within-only correct (b):', in_only)
print('Cross-only correct  (c):', cr_only)
print('Both wrong        :', both_wrong)
print('within acc=%.3f  cross acc=%.3f' % (ok_in.mean(), ok_cr.mean()))
res = mcnemar([[both_ok, in_only],[cr_only, both_wrong]], exact=False, correction=True)
print('McNemar chi2=%.3f  p=%.6g' % (res.statistic, res.pvalue))
print('Manual check chi2 = (|b-c|-1)^2/(b+c) = %.3f' % ((abs(in_only-cr_only)-1)**2/(in_only+cr_only)))
```
